```python
import numpy as np

# Example sentence tokens = 4 words
# Each word embedding dimension = 4 (simplified)
X = np.array([
    [1, 0, 1, 0],   # What
    [0, 1, 0, 1],   # are
    [1, 1, 1, 0],   # symptoms
    [0, 0, 1, 1]    # diabetes
])

# Weight matrices (random for example)
W_Q = np.random.rand(4, 4)
W_K = np.random.rand(4, 4)
W_V = np.random.rand(4, 4)

# Step 1: Compute Q, K, V
Q = X @ W_Q
K = X @ W_K
V = X @ W_V

# Step 2: Compute Attention Scores
# Q × K^T
scores = Q @ K.T

# Step 3: Scaling
d_k = K.shape[1]
scaled_scores = scores / np.sqrt(d_k)

# Step 4: Softmax
def softmax(x):
    return np.exp(x) / np.sum(np.exp(x), axis=-1, keepdims=True)

attention_weights = softmax(scaled_scores)

# Step 5: Final Output
output = attention_weights @ V

print("Attention Scores:\n", scores)
print("\nScaled Scores:\n", scaled_scores)
print("\nAttention Weights:\n", attention_weights)
print("\nFinal Output:\n", output)
```
